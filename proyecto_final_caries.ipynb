# Proyecto final: clasificación de caries registrada

## 1. Problema, objetivo y relación con los entregables

El objetivo es clasificar presencia o ausencia de caries registrada en
smoking.csv. No se predice aparición futura de caries.

Este capítulo amplía el Entregable 1 y el Entregable 2: conserva su EDA,
depuración, partición por perfiles y modelo base, y añade un estimador
implementado, comparación anidada y ablaciones.

- [Entregable 1: datos, EDA y modelo base](entregable_1_clasificacion.ipynb)
- [Entregable 2: informe publicado](https://aduartech81.github.io/MachineLearning/entregas-2/entregable_2.html)

Las tareas de LendingClub y enfermedad cardíaca son actividades distintas.
Sus datos y métricas no se incorporan al experimento de caries.


## 2. Variables, codificación y prevención de fuga

Se conservan 23 predictores; se excluyen ID, oral y tartar.
El diccionario operativo y sus verificaciones pendientes están documentados
en el Entregable 1. Los códigos categóricos se representan mediante one-hot.
Las variables numéricas utilizan bases spline en el modelo propio.

Los 44.582 registros de desarrollo contienen 35.641 perfiles.
Se emplean cinco folds externos y tres internos; las copias de un perfil
permanecen juntas. Imputación, transformaciones y selección de interacciones
se aprenden dentro de cada ajuste.

El test histórico ya fue consultado y no se usa en este nuevo experimento.
La evaluación anidada es interna; no acredita independencia por paciente
ni elimina retrospectivamente las decisiones tomadas en análisis anteriores.


## 3. Modelo implementado, antecedentes e incertidumbre

El modelo suma funciones univariadas spline y hasta cinco productos entre
variables numéricas. Selecciona interacciones mediante estabilidad de
asociaciones residuales en submuestras de perfiles del ajuste.

La pérdida combina clasificación logística, ordenación de pares
positivo–negativo y regularización L2. Se verificó el gradiente mediante
diferencias finitas y se registró convergencia durante los ajustes.

Estos componentes tienen antecedentes. Escribir una implementación propia
no demuestra por sí solo originalidad científica de la integración.
La selección usada es exploratoria y no hereda garantías de control de
falsos descubrimientos de otros métodos de estabilidad.

Los IC son percentiles individuales del bootstrap agrupado, condicionados
a predicciones fijas. No son intervalos de predicción individual ni
desviaciones estándar entre folds. Los contrastes son exploratorios:
no incorporan toda la variabilidad y dependencia del reentrenamiento.

## 4. Utilidad potencial y disponibilidad de información

Un posible uso futuro sería apoyar la priorización de evaluación
odontológica en personas cuyos datos de chequeo ya estuvieran disponibles.
La secuencia temporal de disponibilidad no está acreditada en esta base.

El rendimiento observado y la sensibilidad con el umbral predeterminado
no respaldan uso clínico. El modelo no sustituye exploración odontológica.
Se requieren verificación de procedencia, definición diagnóstica,
evaluación de costos y validación adicional antes de proponer un servicio.

## 6. Conclusiones y respuestas para la defensa

**¿Por qué clasificación?** El objetivo registrado tiene dos clases:
presencia o ausencia de caries.

**¿Por qué esa codificación?** Los códigos categóricos no se tratan
como magnitudes; las transformaciones se aprenden dentro del ajuste.

**¿Qué representan los intervalos?** Incertidumbre condicional de evaluación,
con remuestreo por perfiles. No abarcan toda la variabilidad de entrenamiento.

**¿Qué aporta la implementación?** Un procedimiento explícito de funciones
no lineales, interacciones limitadas y pérdida combinada, evaluado mediante
ablaciones. Su originalidad científica sigue sin demostrarse.

**¿Mejoró el AUC?** No se encontró evidencia de mejora del modelo completo.
No se confunde ausencia de significancia con equivalencia.

**¿Qué utilidad práctica tiene ahora?** Es un prototipo académico evaluado,
sin utilidad clínica acreditada.

## Aporte de los componentes
Las ablaciones no aportan evidencia de beneficio de las interacciones
seleccionadas ni de la pérdida de ranking. La variante aditiva tiene
el mayor AUC medio observado, pero esa posición descriptiva no demuestra
superioridad estadística ni justifica seleccionarla retrospectivamente
como ganadora confirmada.

Las ablaciones comparan procedimientos con ajuste interno de sus
hiperparámetros; no mantienen necesariamente idéntica regularización.

## Alcance de los intervalos y contrastes
Se remuestrearon perfiles completos dentro de cada fold y clase,
con los mismos pesos para todos los modelos. El estadístico es
la media no ponderada de los cinco AUC externos.

Los intervalos son percentiles individuales del 95 %, no simultáneos.
El contraste bootstrap centrado es aproximado y exploratorio.
Los modelos y sus predicciones permanecen fijos: el procedimiento
no incorpora variabilidad de reentrenamiento ni toda la dependencia
inducida por conjuntos de entrenamiento solapados.
La independencia entre perfiles no está verificada por paciente.
No se trata de una evaluación independiente confirmatoria.

## Utilidad y umbral
Con probabilidad 0,5 —o puntuación 0 para SVM— la sensibilidad es
prácticamente nula. La exactitud y especificidad elevadas no compensan
la falta de identificación de casos positivos.
Estos resultados no respaldan uso clínico de detección de caries.

Un umbral alternativo requiere un criterio operativo explícito y
selección exclusivamente dentro del desarrollo interno. No se elegirá
un corte examinando el mejor resultado de las predicciones externas.
Cambiar el umbral no mejora el AUC del mismo conjunto de puntuaciones.

## Originalidad
Se implementó un estimador con bases spline, selección exploratoria
de interacciones y pérdida combinada de clasificación y ranking.
Sus componentes tienen antecedentes. La originalidad científica de
la integración no está demostrada por haber escrito código propio.
No se afirma superioridad ni se garantiza una calificación académica.

## Alcance del problema
El objetivo es caries registrada, no aparición futura.
El test histórico ya fue consultado y no se presenta como nueva
validación independiente. Los resultados actuales proceden de
evaluación interna mediante validación cruzada anidada agrupada.
