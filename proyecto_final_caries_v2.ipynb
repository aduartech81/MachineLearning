# Proyecto final de caries — actualización V2

Andrés Duarte Chaves · Maestría en Ingeniería Biomédica

Esta actualización complementa el capítulo V1 conservado. Corresponde
exclusivamente al proyecto final de caries.

[Aplicación independiente en Render](https://caries-proyecto-final-andres-duarte.onrender.com/)

La versión publicada en Render debe verificarse después del despliegue.
Este documento presenta evidencia experimental exploratoria; no una
herramienta clínica validada.


## 1. Problema y alcance

En este trabajo evalué la clasificación de presencia o ausencia de caries registrada mediante 23 predictores
de chequeo general. No se predice aparición futura de caries.

El desarrollo contiene 44.582 filas y 35.641 perfiles. El test histórico
ya consultado no constituye una nueva evaluación independiente.
El uso potencial de priorización odontológica requiere verificar la
disponibilidad temporal de los predictores y validación adicional.


## 2. Datos y diseño experimental

Conservé las particiones agrupadas: cinco folds externos y tres internos.
Las transformaciones y los balanceos se ajustaron dentro del entrenamiento.
Agrupar perfiles idénticos no acredita independencia por paciente.

V2 incluye siete familias, cuatro optimizadores y cuatro estrategias de
balanceo: sin balanceo, SMOTE, ADASYN y ponderación.
KNN ponderado no es aplicable. Se completaron 108 combinaciones y
540 evaluaciones de folds.

El presupuesto nominal fue de 16 propuestas por búsqueda; las configuraciones
únicas pueden diferir entre optimizadores. Las 108 combinaciones no representan
el número total de candidatos entrenados.

SMOTE y ADASYN en el espacio codificado pueden generar categorías fraccionarias.
La ponderación utiliza pesos de clase o de observación.

Las configuraciones finales se eligieron por AUC interno por familia y fold.
Esta regla se documentó después de consultar resultados V2: el análisis
permanece exploratorio.

## 3. Contribución implementada y ablaciones

Desarrollé un modelo que combina funciones spline, interacciones numéricas limitadas,
pérdida logística y ranking por pares, con regularización L2.
La implementación y sus gradientes fueron verificados. La novedad científica
de la combinación no está establecida.

La selección de interacciones utiliza residuos de ajuste. Las repeticiones
submuestrean perfiles pero reutilizan el modelo preliminar y sus residuos:
no equivalen a reentrenar todo el procedimiento de selección en cada repetición.

Las ablaciones conservadas permiten evaluar variantes sin interacciones,
sin ranking y aditiva. Sus espacios de regularización no son idénticos;
comparan procedimientos ajustados y no aíslan perfectamente cada componente.

## 4. Evaluación e incertidumbre

La calibración sigmoide y el umbral Youden se ajustaron dentro del entrenamiento
externo, utilizando predicciones internas. Youden no expresa por sí solo
costos clínicos ni acredita un umbral de derivación.

Los intervalos siguientes corresponden a la media del AUC de los cinco folds.
Se utilizaron 2.000 réplicas de perfiles completos dentro de cada fold,
sin estratificación por clase y con los mismos pesos para todos los modelos.

Los modelos y las predicciones permanecen fijos. Los intervalos no incorporan
toda la variabilidad del reentrenamiento ni su dependencia por solapamiento.
Son intervalos individuales, no simultáneos. Los diez contrastes bilaterales
aproximados usan bootstrap centrado y corrección Holm.

## Comparación global de las siete referencias

Comparé los rangos del AUC externo de las siete referencias V2 en los cinco folds compartidos. Un rango menor indica mayor AUC dentro del fold. El modelo propio y sus ablaciones conservan sus contrastes específicos: no los incorporé como si tuvieran el mismo presupuesto de búsqueda V2.

Friedman obtuvo χ²(6) = 21,4286 y p asintótico = 0,001536. La diferencia crítica de Nemenyi fue 4,0282 (α = 0,05). Solo logística–GaussianNB (p = 0,021791) y XGBoost–GaussianNB (p = 0,004720) superaron esa diferencia. No detecté diferencias por Nemenyi entre las referencias principales; esto no demuestra equivalencia.

![Rangos medios y diferencia crítica de Nemenyi.](_static/proyecto_final_caries_rangos/diferencia_critica.png)

**Alcance exploratorio.** Hay solo cinco bloques y los entrenamientos se solapan. La aproximación chi-cuadrado de Friedman es poco fiable con tan pocos bloques. Como sensibilidad, 100.000 permutaciones de etiquetas dentro de fold dieron p = 0,000070, pero tampoco corrigen el solapamiento del entrenamiento ni convierten el análisis en confirmatorio. No interpreto estos resultados como evidencia de utilidad clínica.

Nemenyi controla conjuntamente sus 21 pares bajo sus supuestos; esos p-valores no son los p-valores Holm de los diez contrastes bootstrap del modelo propio. Los dos análisis responden a preguntas diferentes.

Fuentes metodológicas: [SciPy, Friedman](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.friedmanchisquare.html) y [Demšar, 2006](https://www.jmlr.org/papers/v7/demsar06a.html). La aplicación clásica compara datasets independientes; aquí su uso sobre folds es una aproximación exploratoria solicitada por la guía.

## 6. Conclusiones y límites

El modelo propio obtuvo AUC medio 0,6051. Las referencias seleccionadas
internamente alcanzaron aproximadamente 0,6059 en logística y 0,6055 en XGBoost.
La discriminación observada es limitada.

El modelo propio mostró una ventaja condicional exploratoria frente a KNN
y GaussianNB, pero no frente a las referencias con mayor AUC. Las ablaciones
no demostraron una ventaja de añadir interacciones o ranking.

Los umbrales internos V2 modificaron el compromiso entre falsos positivos
y negativos. No demuestran una mejora del AUC ni utilidad clínica.
El modelo propio V1 conserva su evaluación con umbral descriptivo 0,5.

LIME presentó menor fidelidad en los casos negativos. En el verdadero negativo
ID 2, su error absoluto medio fue de 6,03 puntos porcentuales. Las explicaciones
no establecen relaciones causales ni estabilidad general.

La procedencia clínica, la definición diagnóstica, la licencia y la disponibilidad
temporal de variables requieren verificación adicional. El test histórico ya
consultado no puede presentarse como una nueva validación independiente.

Una futura selección de interacciones con residuos fuera de muestra sería
una modificación exploratoria que debe documentarse y evaluarse por separado.
No reemplazaría retrospectivamente estos resultados.
