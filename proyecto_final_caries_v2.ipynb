Andrés Duarte Chaves · Maestría en Ingeniería Biomédica

Esta actualización complementa el capítulo V1 conservado. Corresponde
exclusivamente al proyecto final de caries.

[Aplicación independiente en Render](https://caries-proyecto-final-andres-duarte.onrender.com/)

La versión publicada en Render debe verificarse después del despliegue.
Este documento presenta evidencia experimental exploratoria; no una
herramienta clínica validada.


## 1. Problema y alcance

En este trabajo evalué la clasificación de presencia o ausencia de caries registrada mediante 23 predictores
de chequeo general. No se predice aparición futura de caries.

El desarrollo contiene 44.582 filas y 35.641 perfiles. El test histórico
ya consultado no constituye una nueva evaluación independiente.
El uso potencial de priorización odontológica requiere verificar la
disponibilidad temporal de los predictores y validación adicional.


## 2. Datos y diseño experimental

Conservé las particiones agrupadas: cinco folds externos y tres internos.
Las transformaciones y los balanceos se ajustaron dentro del entrenamiento.
Agrupar perfiles idénticos no acredita independencia por paciente.

V2 incluye siete familias, cuatro optimizadores y cuatro estrategias de
balanceo: sin balanceo, SMOTE, ADASYN y ponderación.
KNN ponderado no es aplicable. Se completaron 108 combinaciones y
540 evaluaciones de folds.

El presupuesto nominal fue de 16 propuestas por búsqueda; las configuraciones
únicas pueden diferir entre optimizadores. Las 108 combinaciones no representan
el número total de candidatos entrenados.

SMOTE y ADASYN en el espacio codificado pueden generar categorías fraccionarias.
La ponderación utiliza pesos de clase o de observación.

Las configuraciones finales se eligieron por AUC interno por familia y fold.
Esta regla se documentó después de consultar resultados V2: el análisis
permanece exploratorio.

### Procedimiento de preparación

Importación del archivo → exclusiones justificadas → construcción de perfiles → particiones agrupadas → ajuste de imputación, escalado y codificación dentro del entrenamiento → búsqueda interna → evaluación del fold externo. El balanceo, cuando se aplica, afecta únicamente al entrenamiento interno o al reajuste, nunca al fold de evaluación.

### Referencia trivial para interpretar las métricas

En desarrollo hay 9.478 positivos y 35.104 negativos: predecir siempre ausencia alcanza 78,74 % de exactitud, pero sensibilidad y F1 iguales a cero. Por eso no seleccioné modelos solo por exactitud. Esta referencia describe las etiquetas de desarrollo; no sustituye al Dummy ajustado con prevalencia de entrenamiento en cada fold.

### Qué conservé y qué amplié

Conservé el modelo logístico inicial y su test histórico como antecedentes. La ampliación incorporó el diccionario de 23 predictores, curvas de aprendizaje agrupadas, evaluación anidada, búsqueda de siete referencias, calibración y umbral internos, contrastes y explicaciones locales. Esta lista documenta cambios; no certifica que todas las observaciones originales del docente estén resueltas sin disponer de su lista completa.

### Cómo leer V1 y V2

V1 es un antecedente experimental conservado. Su «Boosting» corresponde a HistGradientBoostingClassifier; XGBoost es una referencia distinta de V2. También difieren los espacios de búsqueda y las familias comparadas. Por ello cambian AUC, desviaciones y número de contrastes. Las tablas V1 usan umbral 0,5 o puntuación cero en SVM; un Brier vacío en SVM significa que no se proporcionaron probabilidades, no un valor cero. Una precisión alta basada en muy pocos positivos predichos no indica alta sensibilidad.

### Excepción de KNN y presupuesto

Las 112 combinaciones nominales de clasificación se reducen a 108 aplicables porque KNN no implementa el mismo mecanismo de ponderación de clases utilizado en las otras familias. `weights="distance"` pondera vecinos por distancia y no equivale a `class_weight`. Las 540 evaluaciones externas son 108 combinaciones por cinco folds; no son 540 entrenamientos totales ni observaciones independientes. El número efectivo de candidatos, reajustes, fallos y reutilizaciones debe recuperarse de los historiales V2.


## 3. Contribución implementada y ablaciones

Desarrollé un modelo que combina funciones spline, interacciones numéricas limitadas,
pérdida logística y ranking por pares, con regularización L2.
La implementación y sus gradientes fueron verificados. La novedad científica
de la combinación no está establecida.

La selección de interacciones utiliza residuos de ajuste. Las repeticiones
submuestrean perfiles pero reutilizan el modelo preliminar y sus residuos:
no equivalen a reentrenar todo el procedimiento de selección en cada repetición.

Las ablaciones conservadas permiten evaluar variantes sin interacciones,
sin ranking y aditiva. Sus espacios de regularización no son idénticos;
comparan procedimientos ajustados y no aíslan perfectamente cada componente.

## Búsquedas V2: presupuesto efectivo, cómputo y reproducción

Recuperé y comprobé por SHA-256 los 35.585 archivos del paquete de ejecución. Separé las 35 predicciones externas de las 35 internas: las internas se usaron para calibrar y fijar el umbral, no para estimar el desempeño externo. Las decisiones finales coinciden con las huellas publicadas y cada familia conserva las mismas 44.582 filas externas, una vez por fila, sin cruce de perfiles entre entrenamiento interno y validación externa.

### Qué ejecuté y qué significa el presupuesto

Completé 540 búsquedas correspondientes a 108 combinaciones y cinco folds externos. Hubo 8.640 propuestas nominales, 7.139 configuraciones distintas al sumar las búsquedas y 21.417 ajustes internos registrados. Una configuración repetida en búsquedas diferentes se cuenta en cada búsqueda. Estos ajustes no incluyen los reajustes para calibración y evaluación final. TPE y genética reutilizaron resultados de propuestas repetidas; las repeticiones consumieron presupuesto, sin un nuevo ajuste. Por ello, igualdad de propuestas no significa igualdad de evaluaciones únicas.

### TPE y búsqueda genética

En TPE utilicé Optuna 5.0.0, 16 propuestas, cuatro ensayos iniciales aleatorios, 24 candidatos de adquisición, `multivariate=False`, `constant_liar=False` y semilla 42 + fold externo. TPE estima una densidad l(x) para configuraciones con mejores resultados y otra g(x) para el resto. Selecciona candidatos con mayor razón l(x)/g(x); bajo su formulación, este criterio está relacionado con maximizar la mejora esperada. Es un estimador de densidades, no un proceso gaussiano. No cambié la regla al consultar el fold externo.

En genética utilicé DEAP: población de cuatro, cuatro descendientes, inicialización y tres generaciones de descendencia hasta 16 propuestas. Conservé los cuatro mejores del conjunto de padres y descendientes; seleccioné progenitores por torneo de tamaño dos. El cruce uniforme tiene probabilidad 0,8 y probabilidad por gen 0,5. La mutación se aplica a cada hijo con probabilidad 0,5 y a cada gen con probabilidad 1/d. Los genes categóricos se remuestrean; los reales se perturban en escala logarítmica con desviación 0,2 del ancho del dominio y se acotan al intervalo permitido. La semilla es 42 + fold. No hice múltiples reinicios por búsqueda: estos resultados no prueban estabilidad general del optimizador.

Grid y Random evaluaron 16 configuraciones distintas por búsqueda. Los dominios y parámetros fijos completos están en el paquete de código. Grid recorre los puntos publicados; Random y los métodos adaptativos pueden usar valores intermedios en dominios logarítmicos. Los dominios son comparables, pero las mallas no son idénticas.

### Cómo interpretar los tiempos

Los tiempos provienen de los registros originales, no de estimaciones retrospectivas. La tabla suma preprocesamiento, ajuste, predicción y tiempos de propuesta registrados. Son componentes observados, no duración de pared de toda la ejecución ni benchmark aislado. En Grid/Random se procesaron candidatos dentro de cada fold interno; el historial por propuesta reorganiza sus tres ajustes y no representa la cronología real. El tiempo de propuesta genética no incluye toda la generación de descendencia. No atribuyo la menor suma temporal de genética a mayor eficiencia intrínseca: evaluó menos configuraciones únicas.

Conservé los tiempos de las 540 evaluaciones externas en otra tabla. El piloto de logística tiene campos diferentes; las celdas no registradas permanecen vacías. No medí RAM máxima durante las ejecuciones originales. Tampoco dispongo todavía de un benchmark emparejado de parámetros por defecto frente a optimizados; estos tiempos no lo sustituyen.

### SVM y coste teórico

Usé LinearSVC con pérdida squared_hinge, dual=False, máximo 3.000 iteraciones y tolerancia 0,0001. Esta elección controla el coste en 44.582 filas y 32 columnas codificadas; no constituye una búsqueda de kernels ni demuestra que RBF sea inferior. Los parámetros fijos y dominios de C quedan publicados.

Como orientación algorítmica, para n filas, p columnas, I iteraciones, T árboles, profundidad h y B bins: GaussianNB recorre O(np); logística y SVM lineal suelen requerir trabajo del orden O(Inp), dependiente del optimizador y su convergencia. KNN por fuerza bruta cuesta O(np) por consulta y almacena O(np). Un árbol balanceado con búsqueda de cortes tiene un coste típico O(p n log n), dependiente de implementación; un bosque lo multiplica por T y las variables examinadas. Boosting por histogramas depende de T, n, p, h y B; no lo reduzco a una constante independiente del número de árboles. Son órdenes orientativos, no mediciones de memoria ni garantías universales del ejecutor. El balanceo aumenta n y puede aumentar tanto tiempo como almacenamiento.

### Calibración antes y después

Comparé las probabilidades originales, la sigmoide principal y la isotónica secundaria sobre las mismas predicciones externas. Los modelos balanceados o ponderados pueden producir probabilidades originales alejadas de la prevalencia real. En logística, Brier pasó de 0,241400 a 0,163357 con sigmoide; en XGBoost, de 0,178529 a 0,163067. KNN cambió poco: 0,164020 a 0,163884. SVM no produce una probabilidad original, por lo que ese valor no aplica. La referencia constante se calculó con la prevalencia del entrenamiento de cada fold.

Una sigmoide creciente conserva el orden dentro de cada fold. Sin embargo, cada fold tiene un calibrador diferente: al reunir sus predicciones, pueden cambiar los ordenamientos entre folds y el AUC OOF agrupado. No interpreto ese cambio como una mejora del AUC dentro de los folds. Un ECE bajo tampoco demuestra buena discriminación. La sigmoide se mantuvo como método principal; la isotónica es una comparación secundaria, no una selección posterior por desempeño externo.

### DeLong y tamaños de efecto: análisis de sensibilidad

Añadí los diez contrastes dirigidos mediante DeLong clásico sobre AUC OOF agrupado, con Holm y BH, y tamaños de efecto descriptivos sobre los cinco AUC de folds. Este DeLong trata las filas como independientes: perfiles repetidos y entrenamiento solapado incumplen ese supuesto. Sus p e intervalos son una sensibilidad metodológica, no evidencia confirmatoria ni sustituto del bootstrap por perfiles. Además, el AUC agrupado y el promedio de AUC por fold son estimandos distintos.

Cliff describe la dominancia marginal entre las dos series de cinco valores y no conserva el pareamiento; por ello también presenté la correlación biserial de rangos pareados y Cohen dz. Con sólo cinco folds dependientes no asigné categorías universales de magnitud ni intervalos inferenciales a estos efectos. La conclusión principal permanece: no hay mejora demostrada del modelo propio frente a logística, SVM o XGBoost en el bootstrap agrupado exploratorio.

### Dos entregas y límites de reproducción

E2 se prepara para el envío de hoy. E3, para mañana, exige además modelos finales serializados, prueba de ida y vuelta, predicción en vivo con explicación local, revisión de novedad y validación del propio. Esos componentes no se dan por resueltos por aparecer aquí. La guía E3, p.19, permite resultados costosos precalculados y versionados si se incluye el código generador; eso no permite afirmar que la reproducción integral ya se ejecutó.

## Fundamento del modelo propio: alcance verificable

La tarea es clasificar caries registrada: $y_i\in\{0,1\}$ y $x_i$ contiene los 23 predictores conservados. No se pronostica incidencia futura. Para una representación aprendida solo con el ajuste, $z_i=\phi(x_i)$, el score es $s_i=b+z_i^\top w$ y la salida inicial es $\sigma(s_i)=1/(1+e^{-s_i})$.

Las numéricas usan splines cúbicos; las categóricas, one-hot. Se consideran hasta cinco productos entre numéricas estandarizadas. La selección por asociación con residuos es una heurística interna: no acredita control de falsos descubrimientos ni estabilidad inferencial.

Para $M$ pares positivos-negativos fijos, la función implementada es:

$$
L(b,w)=(1-\alpha)\frac1n\sum_i\left[\log(1+e^{s_i})-y_i s_i\right]+\frac{\alpha}{M}\sum_{(p,q)}\log(1+e^{-(s_p-s_q)})+\frac{\lambda}{2}\lVert w\rVert_2^2.
$$

La BCE favorece el ajuste binario; el ranking favorece scores mayores en positivos; L2 penaliza los coeficientes, no el intercepto. Las pérdidas se normalizan por separado. La mezcla no garantiza probabilidades calibradas.

Para un par, $d=s_p-s_q$ y $r=\sigma(-d)$: la derivada respecto a $s_p$ es $-r$ y respecto a $s_q$, $+r$. La parte BCE aporta $(\sigma(s_i)-y_i)/n$. El gradiente de $w$ combina $Z^\top$ por las derivadas del score y $\lambda w$. Se entrega a L-BFGS-B; el máximo de iteraciones y el indicador de convergencia se registran.

**Propiedad condicional.** Con representación y pares fijos, $0\leq\alpha<1$, $\lambda>0$ y ambas clases presentes, la pérdida es estrictamente convexa en $(b,w)$. Cada término logístico tiene segunda derivada no negativa; L2 es positiva en cualquier dirección con componente de coeficientes no nula. En la dirección exclusiva del intercepto, la BCE aporta curvatura estrictamente positiva para scores finitos. Ambas clases y L2 impiden escapar al infinito sin aumentar el objetivo, por lo que existe un mínimo único. Esto no convierte en convexa la selección previa de interacciones ni garantiza que el criterio numérico haya alcanzado el óptimo.

**Complejidad.** Con $d$ columnas transformadas, $p$ numéricas, $B$ submuestras, $I_0$ iteraciones preliminares e $I$ finales, el trabajo principal es del orden de $I_0nd+Bn p^2+I(nd+M)$. La matriz de todos los productos introduce memoria del orden de $np^2$; las bases, $nd$. La inferencia de $m$ filas escala con su transformación y $md$. Son órdenes del algoritmo implementado, no una medición de memoria del propio ni un estudio empírico de escalamiento.

**Pseudocódigo.** (1) Validar esquema y dos clases. (2) Aprender imputación, splines, codificación y escalas solo en ajuste. (3) Fijar pares con semilla. (4) Ajustar aditivo preliminar y seleccionar interacciones dentro de ajuste. (5) Minimizar la pérdida con gradiente y registrar convergencia. (6) Transformar nuevas filas con los objetos guardados y calcular score/sigmoide; si se calibra, usar calibrador aprendido con OOF interno.

**Resultado y originalidad.** AUC medio V1 propio: 0,605062; aditivo sin ambos componentes: 0,606151. Las ablaciones no prueban ventaja de añadir ranking/interacciones. Implementación propia y una propiedad matemática verificable no demuestran novedad científica (R1) ni superioridad (R3). La revisión sistemática de novedad sigue pendiente.
